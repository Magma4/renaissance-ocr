# 01 Data Exploration

This notebook is for quick inspection of extracted page images, layout variation, and likely failure cases for the printed OCR baseline.

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

page_dir = Path('../data/page_images')
page_paths = sorted(page_dir.glob('*.png'))
len(page_paths), page_paths[:5]
```

```python
sample_path = page_paths[0]
image = Image.open(sample_path)
plt.figure(figsize=(10, 14))
plt.imshow(image)
plt.axis('off')
plt.title(sample_path.name);
```
